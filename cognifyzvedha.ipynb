LEVEL 2 TASK 1:[RESTAURANT RATINGS]

```python
bins = [0, 1, 2, 3, 4, 5]
labels = ["0–1", "1–2", "2–3", "3–4", "4–5"]
df['Rating Range'] = pd.cut(df['Aggregate rating'], bins=bins, labels=labels, include_lowest=True)
rating_distribution = df['Rating Range'].value_counts().sort_index()
```

```python
print("Rating Range Distribution:")
print(rating_distribution)
```

```python
plt.figure(figsize=(7, 5))
plt.bar(rating_distribution.index.astype(str), rating_distribution.values)
plt.xlabel("Rating Range")
plt.ylabel("Number of Restaurants")
plt.title("Distribution of Aggregate Ratings")
plt.show()
```

```python
mostcom_range = rating_distribution.idxmax()
mostcom_count = rating_distribution.max()
```

```python
print("\nMost Common Rating Range:")
print(f"{mostcom_range} ({mostcom_count} restaurants)")
```

```python
avg_votes = df['Votes'].mean()
print("Average number of votes received by restaurants:", avg_votes)
```

```python
import folium
```

```python
df_map = df.dropna(subset=['Latitude', 'Longitude'])
map_center = [df_map['Latitude'].mean(), df_map['Longitude'].mean()]
restaurant_map = folium.Map(location=map_center, zoom_start=5)
for _, row in df_map.iterrows():
    folium.CircleMarker(
        location=[row['Latitude'], row['Longitude']],
        radius=2,
        color='blue',
        fill=True,
        fill_opacity=0.6
    ).add_to(restaurant_map)

restaurant_map
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
df = pd.read_csv('Dataset .csv')
coords = df[['Latitude', 'Longitude']].dropna()
k = 5
kmeans = KMeans(n_clusters=k, random_state=42)
coords['Cluster'] = kmeans.fit_predict(coords)
plt.figure(figsize=(8,6))
for cluster in range(k):
    subset = coords[coords['Cluster'] == cluster]
    plt.scatter(subset['Longitude'], subset['Latitude'], s=10, label=f"Cluster {cluster}")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Restaurant Location Clusters")
plt.legend()
plt.tight_layout()
plt.show()
```

LEVEL 2 TASK 4:RESTAURANT CHAINS


```python
restaurant_counts = df['Restaurant Name'].value_counts()
chains = restaurant_counts[restaurant_counts > 1]
print("Number of Restaurant Chains:", len(chains))
print("\nSample Restaurant Chains:")
print(chains.head(10))
```

```python
chain_df = df[df['Restaurant Name'].isin(chains.index)]
chain_analysis = chain_df.groupby('Restaurant Name').agg({
    'Aggregate rating': 'mean',
    'Votes': 'mean'
}).round(2)
top_chains = chain_analysis.sort_values(by='Votes', ascending=False).head(10)
top_chains
```

```python
plt.figure(figsize=(10, 5))
plt.bar(top_chains.index, top_chains['Aggregate rating'])
plt.xlabel("Restaurant Chain")
plt.ylabel("Average Rating")
plt.title("Average Ratings of Top Restaurant Chains")
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.bar(top_chains.index, top_chains['Votes'])
plt.xlabel("Restaurant Chain")
plt.ylabel("Average Votes")
plt.title("Popularity of Restaurant Chains")
plt.xticks(rotation=45, ha='right')
plt.show()
```
